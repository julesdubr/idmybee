```python
with open(REFS_IN) as f:
    data = json.load(f)

fig, axes = plt.subplots(n, 2, figsize=(10,n*3))
keep = []
i = 0

for d in data:
    img_path = PROJECT_ROOT / d["image"]
    stem = Path(img_path).stem
    
    if stem in ['FR15ABOM110721013_S2', '2007.0712_S2', 'ABAURA5676_S5']:
        continue

    keep.append({
        "image": str(img_path),
        "boxes": [d["box"]]
    })

    color = cv2.imread(img_path)
    color = cv2.cvtColor(color, cv2.COLOR_BGR2RGB)

    x1,y1,x2,y2 = d["box"]
    axes[i, 1].imshow(color[y1:y2,x1:x2])
    # axes[i, 1].axis('off')

    cv2.rectangle(color, (x1,y1), (x2,y2), (255, 0, 0), 10)

    h, w, _ = color.shape
    scale = 1024 / w
    small = cv2.resize(color, (1024, int(h*scale)), interpolation=cv2.INTER_LINEAR)
    axes[i, 0].imshow(small)
    axes[i, 0].axis('off')
    axes[i, 0].set_title(stem)

    i+=1

plt.tight_layout()
plt.savefig('temp.png', dpi=300)
plt.close()
```

```python
keep_str = json.dumps(keep, indent=2)

with open(REFS_OUT, "w") as f:
    f.write(keep_str)
```
